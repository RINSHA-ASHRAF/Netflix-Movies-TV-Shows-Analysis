## Data Cleaning and Preprocessing

```python
df.isnull().sum()
```

```python
df.isnull().mean()*100
```

```python
columns = ['director', 'cast', 'country', 'rating']

for col in columns:
    df[col] = df[col].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), errors='coerce')
```

```python
df.date_added
```

```python
df['year_added'] = df['date_added'].dt.year
```

```python
df['month_added'] = df['date_added'].dt.month_name()
```

```python
df[['date_added', 'year_added', 'month_added']].head()
```

```python
df[['date_added', 'year_added', 'month_added']].tail()
```

```python
df.columns
```

```python
df.duration
```

```python
df['duration_value'] = pd.to_numeric(
    df['duration'].str.extract(r'(\d+)')[0],
    errors='coerce'
)
```

```python
df['duration_value'].head()
```

```python
df['duration_unit'] = df['duration'].str.extract(
    r'(min|Seasons?)',
    expand=False
)
```

```python
df[['type', 'duration', 'duration_value', 'duration_unit']].head(10)
```

## Exploratory Data Analysis (EDA)

```python
df['type'].value_counts()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='type')

plt.title('Movies vs TV Shows on Netflix')
plt.xlabel('Content Type')
plt.ylabel('Number of Titles')



plt.show()
```

```python
df.columns
```

```python
df.head()
```

```python
# Which years had the highest number of movies and TV shows released
df['release_year'].value_counts().head(10)
```

```python
top_years = df['release_year'].value_counts().head(10)

plt.figure(figsize=(12, 6))

ax = sns.barplot(
    x=top_years.index,
    y=top_years.values
)

plt.title('Top 10 Years by Netflix Content Release')
plt.xlabel('Release Year')
plt.ylabel('Number of Movies or Tv shows')

for container in ax.containers:
    ax.bar_label(container, padding=3)



plt.show()
```

```python
# Which countries contribute the most movies and TV shows to Netflix?
country_df = df.assign(
    country=df['country'].str.split(',')
).explode('country')

country_df['country'] = country_df['country'].str.strip()

top_countries = country_df[
    country_df['country'] != 'Unknown'
]['country'].value_counts().head(10)

top_countries
```

```python
plt.figure(figsize=(12, 6))

ax = sns.barplot(
    x=top_countries.values,
    y=top_countries.index
)

plt.title('Top 10 Countries Producing Netflix Content')
plt.xlabel('Number of Titles')
plt.ylabel('Country')

for container in ax.containers:
    ax.bar_label(container, padding=3)



plt.show()
```

```python
# Which genres are most common on Netflix?
df.columns
```

```python
df['listed_in']
```

```python
# Which genres are most common on Netflix?
genre_df = df.copy()

genre_df['listed_in'] = genre_df['listed_in'].str.split(',')

genre_df = genre_df.explode('listed_in')

genre_df.rename(columns={'listed_in': 'genre'}, inplace=True)
```

```python
top_genres = genre_df['genre'].str.strip().value_counts().head(10)

top_genres
```

```python
plt.figure(figsize=(12, 6))

ax = sns.barplot(
    x=top_genres.values,
    y=top_genres.index
)

plt.title('Top 10 Most Popular Genres on Netflix')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')

for container in ax.containers:
    ax.bar_label(container, padding=3)

plt.savefig('visualizations/top_genres.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python
# Let's find out which age/content ratings are most common on Netflix.
df['rating'].value_counts()
```

```python
rating_counts = df[
    ~df['rating'].isin(['66 min', '74 min', '84 min', 'Unknown'])
]['rating'].value_counts()

rating_counts
```

```python
plt.figure(figsize=(12, 7))

ax = sns.barplot(
    x=rating_counts.values,
    y=rating_counts.index
)

plt.title('Distribution of Content Ratings on Netflix', fontsize=14)
plt.xlabel('Number of Titles')
plt.ylabel('Content Rating')

for container in ax.containers:
    ax.bar_label(container, padding=3)



plt.show()
```

```python
# typical duration of movies available on Netflix.
df['duration_value']
```

```python
movie_df = df[df['type'] == 'Movie']

movie_df['duration_value'].describe()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=movie_df,
    x='duration_value',
    bins=30,
    kde=True
)

plt.title('Distribution of Movie Durations on Netflix')
plt.xlabel('Duration (Minutes)')
plt.ylabel('Number of Movies')



plt.show()
```

```python
# how many seasons Netflix TV shows typically have.
tv_df = df[df['type'] == 'TV Show']

tv_df['duration_value'].describe()
```

```python
plt.figure(figsize=(12, 6))

ax = sns.countplot(
    data=tv_df,
    x='duration_value',
    order=sorted(tv_df['duration_value'].dropna().unique())
)

plt.title('Distribution of Seasons in Netflix TV Shows')
plt.xlabel('Number of Seasons')
plt.ylabel('Number of TV Shows')

for container in ax.containers:
    ax.bar_label(container, padding=3)



plt.show()
```

```python
# Count how many titles Netflix added each year.
yearly_additions = df['year_added'].value_counts().sort_index()

yearly_additions
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    x=yearly_additions.index,
    y=yearly_additions.values,
    marker='o'
)

plt.title('Netflix Content Additions Over the Years')
plt.xlabel('Year Added')
plt.ylabel('Number of Titles')
plt.xticks(yearly_additions.index, rotation=45)
plt.grid(True)


plt.show()
```

```python
# Top 10 directors on Netflix
director_counts = df['director'].value_counts()

director_counts
```

```python
director_df = df.copy()

director_df['director'] = director_df['director'].str.split(',')

director_df = director_df.explode('director')

director_df['director'] = director_df['director'].str.strip()
```

```python
top_directors = director_df[
    director_df['director'] != 'Unknown'
]['director'].value_counts().head(10)

top_directors
```

```python
plt.figure(figsize=(12, 6))

ax = sns.barplot(
    x=top_directors.values,
    y=top_directors.index
)

plt.title('Top 10 Directors on Netflix')
plt.xlabel('Number of Titles')
plt.ylabel('Director')

for container in ax.containers:
    ax.bar_label(container, padding=3)



plt.show()
```

```python
# Compare Movies and TV Shows by year
content_by_year = df.groupby(
    ['release_year', 'type']
).size().reset_index(name='count')

content_by_year.head()
```

```python
plt.figure(figsize=(14, 6))

sns.lineplot(
    data=content_by_year,
    x='release_year',
    y='count',
    hue='type'
)

plt.title('Movies vs TV Shows Released Over the Years')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.legend(title='Content Type')


plt.show()
```


# Final Findings and Conclusions

## 1. Dataset Overview
- The dataset contains 8,807 Netflix titles.
- It includes 6,131 Movies and 2,676 TV Shows.
- Movies account for 69.62% of the content, while TV Shows account for 30.38%.

## 2. Content Release Analysis
- 2018 was among the peak years for Netflix content releases.
- Movies consistently outnumber TV Shows across most release years.
- Both Movies and TV Shows show significant growth in later release years.

## 3. Country Analysis
- The United States contributes the highest number of titles.
- India ranks second, followed by the United Kingdom.
- Several other countries also contribute to Netflix's content library.

## 4. Genre Analysis
- International Movies is the most frequently listed genre.
- Dramas and Comedies are also among the most common genres.
- International TV Shows ranks among the top five genres.

## 5. Content Rating Analysis
- TV-MA is the most common content rating, with 3,207 titles.
- TV-14 ranks second, with 2,160 titles.
- TV-PG ranks third, with 863 titles.

## 6. Duration Analysis
- The average movie duration is approximately 100 minutes.
- The median movie duration is 98 minutes.
- Most TV Shows have two seasons or fewer.

## 7. Content Addition Analysis
- Netflix content additions increased significantly between 2015 and 2019.
- 2019 recorded the highest number of additions, with 2,016 titles.
- A decline in additions is visible from 2020 onwards in this dataset.

## 8. Director Analysis
- Rajiv Chilaka has the highest number of titles in the dataset, with 22.
- Jan Suter and Raúl Campos follow with 21 and 19 titles respectively.

## Overall Conclusion
This exploratory data analysis provides insights into Netflix's content library, including content types, release trends, countries, genres, ratings, durations, and directors.

The analysis demonstrates how Python libraries such as Pandas, NumPy, Matplotlib, and Seaborn can be used to clean, explore, analyze, and visualize real-world datasets.
